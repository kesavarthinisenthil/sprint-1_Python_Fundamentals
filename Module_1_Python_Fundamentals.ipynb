# Variable-Length Arguments

## 1. Understand the Concept

Variable-length arguments allow a function to accept any number of arguments.

`*args` is used for multiple positional arguments.

`**kwargs` is used for multiple keyword arguments.

## 2. Demonstrate the Concept

For example, a function can accept any number of student names or details.

```python
def show_names(*args):
    for name in args:
        print(name)

show_names("Kesavarthini", "Priya", "Anu")


def show_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

show_details(name="Kesavarthini", age=22)
```

## 3. Implementation Explanation

`*args` collects multiple positional arguments into a tuple.

`**kwargs` collects multiple keyword arguments into a dictionary.

This allows functions to accept a flexible number of arguments.

# Recursive Functions

## 1. Understand the Concept

A recursive function is a function that calls itself to solve a problem. It must have a condition to stop the recursion.

## 2. Demonstrate the Concept

For example, we can use recursion to calculate the factorial of a number.

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

## 3. Implementation Explanation

The `factorial()` function calls itself with a smaller value.

When `n` becomes 1, the recursion stops and returns 1.

For `5`, the result is `5 × 4 × 3 × 2 × 1 = 120`.

# Generators

## 1. Understand the Concept

A generator is a special type of function that produces values one at a time instead of storing all values in memory.

## 2. Demonstrate the Concept

For example, a generator can produce numbers one by one when needed.

```python
def numbers():
    for i in range(1, 4):
        yield i

for number in numbers():
    print(number)
```

It produces one number at a time when the `for` loop requests the next value.

# yield

## 1. Understand the Concept

The `yield` keyword is used in a generator to return a value one at a time. It pauses the function and continues from the same point when the next value is requested.

## 2. Demonstrate the Concept

For example, `yield` can produce numbers one by one instead of returning all numbers at once.

```python
def numbers():
    yield 1
    yield 2
    yield 3

for number in numbers():
    print(number)
```

## 3. Implementation Explanation

The function uses `yield` to produce the numbers one at a time.

The `for` loop receives each value and prints it.

# Decorators

## 1. Understand the Concept

A decorator is a function that adds extra functionality to another function without changing its original code.

## 2. Demonstrate the Concept

For example, a decorator can print a message before and after another function runs.

```python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@decorator
def greet():
    print("Hello!")

greet()
```

## 3. Implementation Explanation

The `decorator()` function receives another function and adds extra statements around it.

The `@decorator` syntax applies the decorator to the `greet()` function.

# Closures

## 1. Understand the Concept

A closure is a function that remembers and uses a value from its outer function even after the outer function has finished executing.

## 2. Demonstrate the Concept

For example, an outer function can create a greeting with a specific name, and the inner function can remember that name.

```python
def create_greeting(name):
    def greet():
        print("Hello", name)
    return greet

greeting = create_greeting("Kesavarthini")
greeting()
```

## 3. Implementation Explanation

The `create_greeting()` function receives a name and returns the inner `greet()` function.

The inner function remembers the `name` value even after the outer function has finished.

# global and nonlocal

## 1. Understand the Concept

`global` is used to modify a variable defined outside a function at the global level.

`nonlocal` is used to modify a variable from an outer function inside a nested function.

## 2. Demonstrate the Concept

We can use `global` to change a global variable and `nonlocal` to change a variable in an outer function.

```python
count = 10

def update_global():
    global count
    count = 20

update_global()
print(count)


def outer():
    value = 10

    def inner():
        nonlocal value
        value = 20

    inner()
    print(value)

outer()
```

## 3. Implementation Explanation

`global` allows the function to modify the global variable `count`.

`nonlocal` allows the nested `inner()` function to modify the `value` variable from the `outer()` function.
