# What is Logging?

- Logging is the process of recording information about what happens while a program is running.
- It helps developers monitor the application and understand its behavior.
- Logs can contain information about normal activities, warnings, and errors.
- Logging is useful for debugging and troubleshooting problems in applications.

For example, when a user logs into an application, we can record a log message saying that the login was successful.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in successfully")
```

## Implementation Explanation

I imported Python's built-in `logging` module.

I configured the logging level as `INFO` so that information messages can be displayed.

The `logging.info()` function records a message about the user's successful login.

Logging helps us track what is happening in the application.

## Output

INFO:root:User logged in successfully

# print() vs Logging

- `print()` is mainly used to display information directly on the screen.
- Logging is used to record and monitor application events.
- Logging provides different levels such as INFO, WARNING, and ERROR.
- Logs can also be saved to files, which is useful for debugging and monitoring applications.

For example, `print()` can display a simple message, while logging can record an application event with its log level.

```python
import logging

print("Application started")

logging.basicConfig(level=logging.INFO)
logging.info("Application started")
```

## Implementation Explanation

I used `print()` to display a simple message on the screen.

I used `logging.info()` to record the same event as an informational log.

Logging is more suitable for larger applications because it provides log levels and can save messages to log files.

## Output

Application started

INFO:root:Application started

# Logging Levels

- Logging levels indicate the importance or severity of a log message.
- Python provides levels such as DEBUG, INFO, WARNING, ERROR, and CRITICAL.
- DEBUG is used for detailed information during development.
- INFO is used for normal application activities.
- WARNING indicates a possible problem.
- ERROR indicates that an operation failed.
- CRITICAL indicates a serious problem that may affect the application.

For example, we can use different logging levels to record normal activities, warnings, errors, and critical problems in an application.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debugging information")
logging.info("Application started")
logging.warning("Low memory warning")
logging.error("File processing failed")
logging.critical("Critical system failure")
```

## Implementation Explanation

I set the logging level to `DEBUG` so that all logging levels are displayed.

Each logging function represents a different level of importance.

`DEBUG` gives detailed information, while `CRITICAL` represents the most serious problem.

## Output

DEBUG:root:Debugging information  
INFO:root:Application started  
WARNING:root:Low memory warning  
ERROR:root:File processing failed  
CRITICAL:root:Critical system failure

# Logging Formatters

- A logging formatter controls how log messages are displayed.
- It can include information such as time, log level, and message.
- A consistent format makes logs easier to read and understand.
- Formatters are useful when we need detailed information about application events.

For example, we can format a log message to show the time, logging level, and message.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.warning("Low memory warning")
```

## Implementation Explanation

I used the `format` parameter to define how the log messages should appear.

`%(asctime)s` shows the time, `%(levelname)s` shows the logging level, and `%(message)s` shows the actual message.

This format makes the logs easier to read and understand.

## Output

The exact time will be different when the code runs.

Example:

2026-09-13 11:30:00,000 - INFO - Application started  
2026-09-13 11:30:00,001 - WARNING - Low memory warning

# File Handler

- A file handler is used to send log messages to a file.
- It allows us to store logs instead of displaying them only on the console.
- Saved log files can be checked later for debugging and monitoring.
- File handlers are useful for applications that need a history of events and errors.

For example, we can create an application log file and store information about application activities in it.

```python
import logging

logger = logging.getLogger("application")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

logger.info("Application started")
logger.info("Data processing completed")

print("Logs were written to application.log")
```

## Implementation Explanation

I created a logger and set its level to `INFO`.

I used `FileHandler` to write the log messages into `application.log`.

I used a formatter to give the log entries a consistent format.

This is useful because we can check the log file later to understand what happened in the application.

## Output

Console:

Logs were written to application.log

The `application.log` file will contain entries similar to:

2026-09-13 11:30:00,000 - INFO - Application started  
2026-09-13 11:30:00,001 - INFO - Data processing completed

# Console Handler

- A console handler sends log messages to the console or terminal.
- It allows us to see application events while the program is running.
- We can control which logging levels are displayed on the console.
- It is useful for monitoring an application during development and execution.

For example, we can create a console handler that displays INFO-level messages on the screen.

```python
import logging

logger = logging.getLogger("console_logger")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
logger.addHandler(console_handler)

logger.info("Application started")
logger.warning("This is a warning")
```

## Implementation Explanation

I created a logger and added a `StreamHandler` to send log messages to the console.

I set the handler level to `INFO`, so INFO and higher-level messages are displayed.

I also used a formatter to make the messages easier to read.

## Output

INFO - Application started  
WARNING - This is a warning

# Rotating Logs

- Rotating logs means creating new log files when the current log file reaches a specific size or time limit.
- It prevents a single log file from becoming too large.
- Older log files can be kept for future reference.
- It is useful for applications that generate a large number of log messages.

For example, we can limit a log file to a small size and automatically create another file when the limit is reached.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_logger")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "application.log",
    maxBytes=1000,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

for i in range(100):
    logger.info(f"Processing record {i}")

print("Logs were written with rotation enabled.")
```

## Implementation Explanation

I used `RotatingFileHandler` to control the size of the log file.

`maxBytes=1000` means a new log file can be created when the file reaches approximately 1000 bytes.

`backupCount=3` keeps up to three backup log files.

This prevents the log file from growing continuously and taking too much storage.

## Output

Console:

Logs were written with rotation enabled.

The folder may contain:

- application.log
- application.log.1
- application.log.2
- application.log.3

# Logging Best Practices

- Use meaningful log messages that clearly describe what happened.
- Use the correct logging level based on the importance of the event.
- Avoid logging sensitive information such as passwords or private data.
- Store logs in files when we need to monitor or troubleshoot an application later.
- Use consistent formatting so that logs are easy to read and analyze.

For example, we can log the start of a process using `INFO` and log an error using `ERROR`, without exposing sensitive information.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

username = "Kesavarthini"

logging.info(f"User {username} started the application")

try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("Division failed because the divisor was zero")
```

## Implementation Explanation

I used `INFO` to record a normal application activity.

I used `ERROR` to record a problem that occurred during execution.

I used a consistent format so that the logs are easy to understand.

I did not log sensitive information such as passwords.

## Output

The exact timestamp will be different when the code runs.

Example:

2026-09-13 11:30:00,000 - INFO - User Kesavarthini started the application  
2026-09-13 11:30:00,001 - ERROR - Division failed because the divisor was zero
