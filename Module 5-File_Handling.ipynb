# Reading a Text File

- File handling is used to create, read, write, and manage files using Python.
- Reading a file means getting the data stored inside the file.
- Python provides the `open()` function to work with files.
- We use the `"r"` mode when we want to read an existing text file.

For example, we can create a text file containing a message and then read its contents using Python.

```python
with open("sample.txt", "w") as file:
    file.write("Welcome to Python file handling.")

with open("sample.txt", "r") as file:
    content = file.read()

print(content)
```

## Implementation Explanation

First, I created `sample.txt` and wrote a message into it.

Then I opened the file in `"r"` mode to read its contents.

I used `with` so that Python automatically closes the file after the operation.

The `read()` function reads the complete content of the file.

## Output

Welcome to Python file handling.

# Writing to a Text File

- Writing to a file means storing data inside a file.
- Python uses `"w"` mode to write data to a file.
- If the file does not exist, Python creates it.
- If the file already contains data, `"w"` mode replaces the existing content.

For example, we can store a student's name and course information in a text file.

```python
with open("student.txt", "w") as file:
    file.write("Name: Kesavarthini\n")
    file.write("Course: Artificial Intelligence and Machine Learning")

print("Data written successfully.")
```

## Implementation Explanation

I opened `student.txt` in `"w"` mode to write data into the file.

The `write()` function stores the given text in the file.

I used `\n` to write the course information on a new line.

The `with` statement automatically closes the file after writing.

## Output

Data written successfully.

The `student.txt` file contains:

Name: Kesavarthini  
Course: Artificial Intelligence and Machine Learning

# Appending to a Text File

- Appending means adding new data to the existing content of a file.
- Python uses `"a"` mode for appending data.
- The existing content is not removed when we use append mode.
- We use it when we want to add new information to an existing file, such as logs or records.

For example, we can add a new skill to an existing student file without deleting the information already stored.

```python
with open("student.txt", "a") as file:
    file.write("\nSkill: Python")

print("New data appended successfully.")
```

## Implementation Explanation

I opened `student.txt` using `"a"` mode.

The `write()` function adds the new skill at the end of the existing file.

The previous content remains unchanged.

## Output

New data appended successfully.

The file now contains:

Name: Kesavarthini  
Course: Artificial Intelligence and Machine Learning  
Skill: Python

# with Statement

- The `with` statement is used to work with files safely.
- It automatically closes the file after the operation is completed.
- It helps avoid forgetting to close an opened file.
- It makes file-handling code shorter and easier to manage.

For example, we can use `with open()` to read a file. After the block finishes, Python automatically closes the file.

```python
with open("student.txt", "r") as file:
    content = file.read()
    print(content)

print("File operation completed.")
```

## Implementation Explanation

I used the `with` statement to open and read the file.

The `read()` function gets the content from the file.

After leaving the `with` block, Python automatically closes the file.

This makes file handling safer and prevents resources from remaining open.

## Output

Name: Kesavarthini  
Course: Artificial Intelligence and Machine Learning  
Skill: Python  

File operation completed.

# Reading a CSV File

- CSV stands for Comma-Separated Values.
- CSV files are commonly used to store tabular data such as names, ages, and marks.
- Python provides the `csv` module to read and work with CSV files.
- We use CSV handling when structured data needs to be stored or processed in a simple file format.

For example, we can create a CSV file containing student details and read each row using Python.

```python
import csv

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Age", "Course"])
    writer.writerow(["Kesavarthini", 22, "AIML"])
    writer.writerow(["Anu", 21, "CSE"])

with open("students.csv", "r", newline="") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

## Implementation Explanation

I used the `csv` module to create and read a CSV file.

The `csv.writer()` writes rows into the file.

The `csv.reader()` reads each row from the CSV file.

This is useful when we need to work with structured tabular data.

## Output

['Name', 'Age', 'Course']  
['Kesavarthini', '22', 'AIML']  
['Anu', '21', 'CSE']

# Reading a JSON File

- JSON stands for JavaScript Object Notation.
- JSON is commonly used to store and exchange structured data.
- Python provides the built-in `json` module to work with JSON data.
- We use JSON when data needs to be stored in a structured format, especially for applications and APIs.

For example, we can store student information in a JSON file and then read it using Python.

```python
import json

student = {
    "name": "Kesavarthini",
    "age": 22,
    "course": "AIML"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

with open("student.json", "r") as file:
    data = json.load(file)

print(data)
```

## Implementation Explanation

I created a dictionary containing student information.

I used `json.dump()` to store the dictionary as JSON data in a file.

I used `json.load()` to read the JSON data back into Python.

This is useful for storing structured data in a format that is easy for applications to read and exchange.

## Output

{'name': 'Kesavarthini', 'age': 22, 'course': 'AIML'}

# Pickle

- Pickle is a Python module used to serialize and deserialize Python objects.
- Serialization means converting a Python object into a format that can be stored.
- Deserialization means loading the stored object back into Python.
- We use Pickle when we need to save and restore Python objects such as lists, dictionaries, or custom objects.

For example, we can store a Python dictionary in a file using Pickle and then load it back into Python.

```python
import pickle

student = {
    "name": "Kesavarthini",
    "age": 22,
    "course": "AIML"
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

## Implementation Explanation

I created a dictionary containing student information.

I used `pickle.dump()` to serialize and store the dictionary in a binary file.

I used `pickle.load()` to deserialize the stored data and load it back into Python.

This is useful when we need to save Python objects and restore them later.

## Output

{'name': 'Kesavarthini', 'age': 22, 'course': 'AIML'}

# pathlib

- `pathlib` is a Python module used to work with file and directory paths.
- It provides an object-oriented way to handle paths.
- It makes file and directory operations easier and more readable.
- We can use it to check whether files exist, create files, and work with directories.

For example, we can use `Path` to create a file and check whether the file exists.

```python
from pathlib import Path

file_path = Path("sample_path.txt")

file_path.write_text("Learning pathlib in Python.")

print("File exists:", file_path.exists())
print("File name:", file_path.name)
```

## Implementation Explanation

I used `Path()` to create a file path.

The `write_text()` method writes text into the file.

The `exists()` method checks whether the file exists.

The `name` property gives the name of the file.

`pathlib` makes file path operations easier and more readable.

## Output

File exists: True  
File name: sample_path.txt

# os Module

- The `os` module provides functions to interact with the operating system.
- It can be used to work with files, directories, and paths.
- It helps us perform operating system-related tasks using Python.
- We can use it to check directories, create folders, and get the current working directory.

For example, we can use the `os` module to find the current working directory and check whether a directory exists.

```python
import os

current_directory = os.getcwd()

print("Current directory:", current_directory)
print("Directory exists:", os.path.exists(current_directory))
```

I used `os.path.exists()` to check whether the directory exists.

The `os` module is useful when we need to interact with the operating system from Python.

# shutil Module

- `shutil` is a Python module used for high-level file and directory operations.
- It can copy, move, rename, and remove files or directories.
- It is useful when we need to manage files and folders programmatically.
- It works with the operating system's file system.

For example, we can create a file and use `shutil.copy()` to create a copy of that file.

```python
import shutil

with open("original.txt", "w") as file:
    file.write("This is the original file.")

shutil.copy("original.txt", "copy.txt")

print("File copied successfully.")
```

## Implementation Explanation

I created an `original.txt` file containing some text.

I used `shutil.copy()` to copy the file and create `copy.txt`.

This is useful when we need to create backups or move/copy files programmatically.

## Output

File copied successfully.

A new file named `copy.txt` is created in the same directory.

# File Validation

- File validation means checking whether a file exists and is in the expected format.
- It helps prevent errors before reading or processing a file.
- We can check the file path, file existence, and file extension.
- It is useful when working with input files in data processing and AI/ML projects.

For example, before reading a CSV file, we can check whether the file exists and whether its extension is `.csv`.

```python
from pathlib import Path

file_path = Path("students.csv")

if file_path.exists() and file_path.suffix == ".csv":
    print("Valid CSV file.")
else:
    print("Invalid or missing CSV file.")
```

## Implementation Explanation

- `Path()` creates a file path object.
- `exists()` checks whether the file is available.
- `suffix` checks the file extension.
- The condition confirms that the file exists and is a CSV file.
- This validation helps avoid processing an incorrect or missing file.

## Output

Valid CSV file.

# Directory Operations

- Directory operations are used to create, access, list, and manage folders.
- Python provides modules like `pathlib` and `os` to work with directories.
- We can create a directory and check whether it exists.
- Directory operations are useful when organizing files in applications and data projects.

For example, we can create a folder and check whether the folder exists.

```python
from pathlib import Path

folder = Path("student_files")

folder.mkdir(exist_ok=True)

print("Directory exists:", folder.exists())
print("Directory name:", folder.name)
```

## Implementation Explanation

- `Path()` creates a path for the directory.
- `mkdir()` creates the directory.
- `exist_ok=True` prevents an error if the directory already exists.
- `exists()` checks whether the directory was created.
- `name` gives the directory name.

## Output

Directory exists: True
Directory name: student_files
