# Class

## 1. Understand the Concept

A class is a blueprint or template used to create objects.

It defines the data and behavior that the objects can have.

## 2. Demonstrate the Concept

For example, a `Student` class can define the name and age of a student.

```python
class Student:
    name = "Kesavarthini"
    age = 22

student1 = Student()

print(student1.name)
print(student1.age)
```

## 3. Implementation Explanation

`class Student:` creates a class named `Student`.

The class contains `name` and `age` as attributes.

`student1 = Student()` creates an object from the `Student` class.

We use the object to access the class attributes.

# Object

## 1. Understand the Concept

An object is an instance of a class.

It is used to access the data and methods defined inside the class.

## 2. Demonstrate the Concept

For example, if `Student` is a class, `student1` can be an object created from that class.

```python
class Student:
    name = "Kesavarthini"
    age = 22

student1 = Student()

print(student1.name)
print(student1.age)
```

## 3. Implementation Explanation

`Student` is the class.

`student1 = Student()` creates an object of the `Student` class.

The object `student1` is used to access the `name` and `age` attributes.

# __init__()

## 1. Understand the Concept

`__init__()` is a special method that is automatically called when an object is created.

It is commonly used to initialize the object's attributes.

## 2. Demonstrate the Concept

For example, when creating a student object, we can use `__init__()` to assign the student's name and age.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Kesavarthini", 22)

print(student1.name)
print(student1.age)
```

## 3. Implementation Explanation

`__init__()` is called automatically when `student1` is created.

`self.name` and `self.age` store the values passed to the object.

The object can then access these values using `student1.name` and `student1.age`.

# Instance Variables

## 1. Understand the Concept

Instance variables are variables that belong to a specific object.

Each object can have its own different values for instance variables.

## 2. Demonstrate the Concept

For example, different student objects can have different names and ages.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Kesavarthini", 22)
student2 = Student("Rahul", 23)

print(student1.name, student1.age)
print(student2.name, student2.age)
```

## 3. Implementation Explanation

`self.name` and `self.age` are instance variables.

Each object has its own values for these variables.

`student1` and `student2` have different names and ages.

# Class Variables

## 1. Understand the Concept

A class variable is a variable that is shared by all objects of a class.

It is defined inside the class but outside the methods.

## 2. Demonstrate the Concept

For example, all students can belong to the same college, so `college` can be a class variable.

```python
class Student:
    college = "HICET"

    def __init__(self, name):
        self.name = name

student1 = Student("Kesavarthini")
student2 = Student("Rahul")

print(student1.name, student1.college)
print(student2.name, student2.college)
```

## 3. Implementation Explanation

`college` is a class variable because it is defined inside the class and outside the `__init__()` method.

It is shared by all objects of the `Student` class.

Both `student1` and `student2` access the same `college` value.

# Instance Method

## 1. Understand the Concept

An instance method is a method that works with the data of a specific object.

It takes `self` as its first parameter.

## 2. Demonstrate the Concept

For example, a student object can have a method to display the student's name.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print("Student Name:", self.name)

student1 = Student("Kesavarthini")

student1.display_name()
```

## 3. Implementation Explanation

`display_name()` is an instance method because it uses `self`.

`self.name` accesses the name of the particular object.

`student1.display_name()` calls the method for the `student1` object.

# Class Method

## 1. Understand the Concept

A class method is a method that works with the class itself rather than a specific object.

It uses `@classmethod` and takes `cls` as its first parameter.

## 2. Demonstrate the Concept

For example, a class method can be used to display a value shared by the entire class.

```python
class Student:
    college = "HICET"

    @classmethod
    def display_college(cls):
        print("College:", cls.college)

Student.display_college()
```

## 3. Implementation Explanation

`@classmethod` makes `display_college()` a class method.

`cls` refers to the class itself.

`cls.college` accesses the class variable.

The method can be called directly using the class name.

# Static Method

## 1. Understand the Concept

A static method is a method that does not depend on an object or class data.

It uses the `@staticmethod` decorator and does not require `self` or `cls`.

## 2. Demonstrate the Concept

For example, a method that checks whether a number is positive can be a static method.

```python
class Student:
    
    @staticmethod
    def is_positive(number):
        return number > 0

result = Student.is_positive(10)

print(result)
```

## 3. Implementation Explanation

`@staticmethod` makes `is_positive()` a static method.

The method does not use `self` or `cls`.

It simply receives a number and checks whether it is positive.

# @property

## 1. Understand the Concept

`@property` is a decorator that allows a method to be accessed like an attribute.

It is commonly used to control how an attribute is accessed.

## 2. Demonstrate the Concept

For example, we can create a property to calculate the full name of a student.

```python
class Student:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return self.first_name + " " + self.last_name

student1 = Student("Kesavarthini", "S")

print(student1.full_name)
```

## 3. Implementation Explanation

`@property` allows the `full_name()` method to be accessed like an attribute.

We use `student1.full_name` instead of calling it with parentheses.

The property combines the first name and last name and returns the full name.

# Encapsulation

## 1. Understand the Concept

Encapsulation means bundling data and methods together inside a class.

It also helps control direct access to the data by using access modifiers such as private attributes.

## 2. Demonstrate the Concept

For example, a bank account balance can be kept private and accessed through a method.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

print(account.get_balance())
```

## 3. Implementation Explanation

`__balance` is a private attribute.

It cannot be accessed directly using the normal attribute name.

The `get_balance()` method provides controlled access to the balance.

This demonstrates encapsulation by keeping the data inside the class and controlling how it is accessed.

# Abstraction

## 1. Understand the Concept

Abstraction means hiding the internal implementation details and showing only the necessary features.

In Python, abstraction can be implemented using abstract classes and methods.

## 2. Demonstrate the Concept

For example, a user can call a `start()` method without knowing the internal details of how a car starts.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):

    def start(self):
        print("Car started")

car = Car()
car.start()
```

## 3. Implementation Explanation

`Vehicle` is an abstract class that inherits from `ABC`.

`@abstractmethod` defines a method that must be implemented by the child class.

`Car` provides the implementation of the `start()` method.

The user only needs to call `car.start()` without knowing the internal implementation.

# Inheritance

## 1. Understand the Concept

Inheritance allows a child class to acquire the attributes and methods of a parent class.

It helps in code reusability and avoids writing the same code again.

## 2. Demonstrate the Concept

For example, a `Dog` class can inherit common behavior from an `Animal` class.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog = Dog()

dog.eat()
dog.bark()
```

## 3. Implementation Explanation

`Animal` is the parent class.

`Dog` is the child class that inherits from `Animal`.

The `Dog` object can use both the inherited `eat()` method and its own `bark()` method.

# Single Inheritance

## 1. Understand the Concept

Single inheritance means a child class inherits from only one parent class.

## 2. Demonstrate the Concept

For example, a `Dog` class can inherit from one parent class called `Animal`.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog = Dog()

dog.eat()
dog.bark()
```

## 3. Implementation Explanation

`Animal` is the parent class and `Dog` is the child class.

`Dog(Animal)` means that `Dog` inherits from `Animal`.

Since `Dog` has only one parent class, this is called single inheritance.

# Multiple Inheritance

## 1. Understand the Concept

Multiple inheritance means a child class inherits from more than one parent class.

## 2. Demonstrate the Concept

For example, a `Child` class can inherit features from both `Father` and `Mother`.

```python
class Father:
    def father_skill(self):
        print("Father's skill")

class Mother:
    def mother_skill(self):
        print("Mother's skill")

class Child(Father, Mother):
    pass

child = Child()

child.father_skill()
child.mother_skill()
```

## 3. Implementation Explanation

`Father` and `Mother` are the parent classes.

`Child` inherits from both classes using `Child(Father, Mother)`.

Therefore, the `Child` object can access methods from both parent classes.

# Multilevel Inheritance

## 1. Understand the Concept

Multilevel inheritance means a class inherits from another class, which itself inherits from another class.

It creates a chain of inheritance.

## 2. Demonstrate the Concept

For example, `Animal` is inherited by `Dog`, and `Dog` is inherited by `Puppy`.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

class Puppy(Dog):
    def play(self):
        print("Puppy is playing")

puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.play()
```

## 3. Implementation Explanation

`Dog` inherits from `Animal`.

`Puppy` inherits from `Dog`.

Therefore, the `Puppy` object can access methods from both `Dog` and `Animal`.

This forms a chain of inheritance and is called multilevel inheritance.

# Hierarchical Inheritance

## 1. Understand the Concept

Hierarchical inheritance means multiple child classes inherit from the same parent class.

## 2. Demonstrate the Concept

For example, both `Dog` and `Cat` can inherit common behavior from the `Animal` class.

```python
class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

class Cat(Animal):
    def meow(self):
        print("Cat is meowing")

dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
```

## 3. Implementation Explanation

`Animal` is the parent class.

`Dog` and `Cat` are two child classes that inherit from `Animal`.

Both child classes can use the `eat()` method from the parent class.

This is called hierarchical inheritance.

# Hybrid Inheritance

## 1. Understand the Concept

Hybrid inheritance is a combination of two or more types of inheritance.

It can combine patterns such as multiple inheritance and multilevel inheritance.

## 2. Demonstrate the Concept

For example, `Child` inherits from both `Father` and `Mother`, while `Father` inherits from `Grandparent`.

```python
class Grandparent:
    def family_name(self):
        print("Family name")

class Father(Grandparent):
    def father_skill(self):
        print("Father's skill")

class Mother:
    def mother_skill(self):
        print("Mother's skill")

class Child(Father, Mother):
    def child_skill(self):
        print("Child's skill")

child = Child()

child.family_name()
child.father_skill()
child.mother_skill()
child.child_skill()
```

## 3. Implementation Explanation

`Father` inherits from `Grandparent`, creating multilevel inheritance.

`Child` inherits from both `Father` and `Mother`, creating multiple inheritance.

The combination of these inheritance patterns is called hybrid inheritance.

# Polymorphism

## 1. Understand the Concept

Polymorphism means one method or function can behave differently depending on the object using it.

It allows different classes to use the same method name with different implementations.

## 2. Demonstrate the Concept

For example, different animals can have different sounds using the same `sound()` method.

```python
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

## 3. Implementation Explanation

Both `Dog` and `Cat` have a method named `sound()`.

The method behaves differently depending on the object.

`dog.sound()` prints the dog's sound, while `cat.sound()` prints the cat's sound.

This is an example of polymorphism.

# Method Overriding

## 1. Understand the Concept

Method overriding occurs when a child class provides its own implementation of a method that is already defined in the parent class.

## 2. Demonstrate the Concept

For example, a parent class can have a `sound()` method, and the child class can redefine it with its own behavior.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

animal = Animal()
dog = Dog()

animal.sound()
dog.sound()
```

## 3. Implementation Explanation

`Animal` defines the `sound()` method.

`Dog` inherits from `Animal` but provides its own version of `sound()`.

When `dog.sound()` is called, the child class version is executed.

This is called method overriding.

# Method Overloading

## 1. Understand the Concept

Method overloading means having the same method name with different parameters.

Python does not support traditional method overloading like some other programming languages.

It can be achieved using default arguments or variable-length arguments.

## 2. Demonstrate the Concept

For example, one method can accept one or two numbers and perform addition.

```python
class Calculator:
    def add(self, a, b=0):
        return a + b

calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
```

## 3. Implementation Explanation

The `add()` method uses a default value for `b`.

When only one value is given, `b` is automatically set to `0`.

When two values are given, both values are added.

This is one way to achieve method overloading in Python.

# Composition

## 1. Understand the Concept

Composition is a relationship where one class contains an object of another class.

It represents a strong "has-a" relationship.

## 2. Demonstrate the Concept

For example, a `Car` has an `Engine`. The `Car` class can create and use an `Engine` object.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")

car = Car()
car.start_car()
```

## 3. Implementation Explanation

`Engine` is a separate class.

The `Car` class creates an `Engine` object inside its `__init__()` method.

The `Car` object uses the `Engine` object to start the car.

This strong "has-a" relationship is called composition.

# Aggregation

## 1. Understand the Concept

Aggregation is a relationship where one class uses an object of another class.

It represents a weak "has-a" relationship, where the objects can exist independently.

## 2. Demonstrate the Concept

For example, a `Department` can have a `Teacher`, but the teacher can exist independently of the department.

```python
class Teacher:
    def teach(self):
        print("Teacher is teaching")

class Department:
    def __init__(self, teacher):
        self.teacher = teacher

    def start_class(self):
        self.teacher.teach()

teacher = Teacher()
department = Department(teacher)

department.start_class()
```

## 3. Implementation Explanation

The `Teacher` object is created separately from the `Department` object.

The `Department` receives the existing `Teacher` object.

The teacher can exist independently even if the department object is removed.

This weak "has-a" relationship is called aggregation.

# Association

## 1. Understand the Concept

Association is a relationship where two independent classes interact with each other.

Both objects can exist independently.

## 2. Demonstrate the Concept

For example, a `Student` can interact with a `Teacher` during a class.

```python
class Teacher:
    def teach(self):
        print("Teacher is teaching")

class Student:
    def learn(self):
        print("Student is learning")

teacher = Teacher()
student = Student()

teacher.teach()
student.learn()
```

## 3. Implementation Explanation

`Teacher` and `Student` are independent classes.

Their objects can exist separately.

They are related because they interact in a learning environment.

This type of relationship between independent objects is called association.

# Dunder Methods

## 1. Understand the Concept

Dunder methods are special methods in Python that start and end with double underscores.

They are used to define special behavior for objects.

## 2. Demonstrate the Concept

For example, `__str__()` defines what should be displayed when an object is printed.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"

student = Student("Kesavarthini")

print(student)
```

## 3. Implementation Explanation

`__init__()` is used to initialize the object.

`__str__()` is a dunder method that defines the string representation of the object.

When `print(student)` is executed, Python automatically calls `__str__()`.

# super()

## 1. Understand the Concept

`super()` is used to access methods or attributes from a parent class.

It is commonly used when a child class needs to use the functionality of its parent class.

## 2. Demonstrate the Concept

For example, a child class can use the parent class `__init__()` method using `super()`.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age

student = Student("Kesavarthini", 22)

print(student.name)
print(student.age)
```

## 3. Implementation Explanation

`Student` inherits from `Person`.

`super().__init__(name)` calls the parent class `__init__()` method.

This allows the child class to reuse the parent's initialization logic.

The `Student` class then initializes its own `age` attribute.

# __repr__()

## 1. Understand the Concept

`__repr__()` is a dunder method used to define the developer-friendly string representation of an object.

It is mainly useful for debugging and understanding the object's details.

## 2. Demonstrate the Concept

For example, we can use `__repr__()` to display the name and age of a student object.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"

student = Student("Kesavarthini", 22)

print(repr(student))
```

## 3. Implementation Explanation

`__repr__()` defines how the object is represented when `repr()` is used.

It returns a string containing the student's name and age.

This representation is useful for debugging and inspecting objects.

# __len__()

## 1. Understand the Concept

`__len__()` is a dunder method used to define the behavior of the `len()` function for an object.

It should return the number of items or elements in the object.

## 2. Demonstrate the Concept

For example, we can create a class that stores a list of students and use `len()` to find the number of students.

```python
class Classroom:
    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)

classroom = Classroom(["Kesavarthini", "Rahul", "Priya"])

print(len(classroom))
```

## 3. Implementation Explanation

`__len__()` defines how the `len()` function works for the `Classroom` object.

It returns the number of students in the list.

When `len(classroom)` is called, Python automatically calls `__len__()`.

# Dataclasses

## 1. Understand the Concept

A dataclass is a special type of class that makes it easier to store data.

The `@dataclass` decorator automatically provides methods such as `__init__()` and `__repr__()`.

## 2. Demonstrate the Concept

For example, a `Student` dataclass can store a student's name and age without writing `__init__()` manually.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int

student = Student("Kesavarthini", 22)

print(student)
```

## 3. Implementation Explanation

`@dataclass` automatically creates the `__init__()` method for the class.

We only need to define the attributes and their types.

It also provides a useful `__repr__()` method for displaying the object.

# Singleton Pattern

## 1. Understand the Concept

The Singleton pattern ensures that only one object of a class is created.

It is useful when only one shared instance is needed throughout a program.

## 2. Demonstrate the Concept

For example, we can create a class that allows only one configuration object.

```python
class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```

## 3. Implementation Explanation

`_instance` stores the single object created by the class.

`__new__()` checks whether an instance already exists.

If no instance exists, it creates one. Otherwise, it returns the existing instance.

Therefore, `object1` and `object2` refer to the same object.
