# try and except

- `try` is used to write code that may cause an error.
- `except` is used to handle the error without stopping the entire program.
- It helps the program continue running when an expected error occurs.
- We use it when we want to handle possible runtime errors safely.

For example, when dividing two numbers, the user may enter 0 as the denominator.
We can use `try` and `except` to handle the division error.

```python
try:
    number = 10
    divisor = 0
    result = number / divisor
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

## Implementation Explanation

I used `try` for the division operation because it may cause an error.

The `except` block handles the `ZeroDivisionError` when the divisor is 0.

This prevents the program from stopping suddenly and gives a meaningful message to the user.

## Output

Cannot divide by zero

# else

- `else` is used with `try` and `except`.
- The `else` block runs only when no exception occurs in the `try` block.
- It helps separate the successful code from the error-handling code.
- We use it when we want to perform an action only if the `try` block executes successfully.

For example, we can divide two numbers.
If the division is successful, the `else` block displays the result.
If an error occurs, the `except` block handles it.

```python
try:
    number = 10
    divisor = 2
    result = number / divisor
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Result:", result)
```

## Implementation Explanation

I used `try` for the division operation because it may cause an error.

The `except` block handles the error if the divisor is zero.

Since the division is successful, the `else` block runs and displays the result.

## Output

Result: 5.0

# finally

- `finally` is used with `try` and `except`.
- The `finally` block always executes whether an exception occurs or not.
- It is mainly used for cleanup activities.
- We can use it to close files, database connections, or release resources.

For example, after performing a calculation, we can use `finally` to display a message that the operation is completed.

```python
try:
    number = 10
    divisor = 2
    result = number / divisor
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Operation completed")
```

## Implementation Explanation

I used the `try` block for the division operation.

The `except` block handles the error if the divisor is zero.

The `finally` block runs after the operation, whether an error occurs or not.

## Output

Result: 5.0  
Operation completed

# raise

- `raise` is used to manually generate an exception.
- It allows us to stop the normal flow of a program when a specific condition is not valid.
- We can raise built-in exceptions such as `ValueError` or `TypeError`.
- It is useful for validating user input and enforcing rules in a program.

For example, we can check a person's age.
If the age is less than 18, we can manually raise a `ValueError`.

```python
age = 15

try:
    if age < 18:
        raise ValueError("Age must be 18 or above")
    print("Eligible")
except ValueError as error:
    print(error)
```

## Implementation Explanation

I used `raise` to manually create a `ValueError` when the age is below 18.

The `if` condition checks whether the age meets the required condition.

The `except` block catches the raised error and displays the message.

## Output

Age must be 18 or above

# assert

- `assert` is used to check whether a condition is true.
- If the condition is `True`, the program continues normally.
- If the condition is `False`, Python raises an `AssertionError`.
- It is mainly used for testing assumptions and checking conditions during development.

For example, we can check whether a person's age is valid before continuing the program.

```python
age = 22

try:
    assert age >= 18, "Age must be 18 or above"
    print("Age is valid")
except AssertionError as error:
    print(error)
```

## Implementation Explanation

I used `assert` to check whether the age is 18 or above.

Since the condition is `True`, no `AssertionError` occurs and the program continues.

The `except` block would handle the error if the condition was `False`.

## Output

Age is valid

# Custom Exceptions

- A custom exception is an exception created by the programmer for a specific situation.
- It is created by defining a class that inherits from `Exception`.
- It makes error messages more meaningful for a particular application.
- We use custom exceptions when built-in exceptions do not clearly describe the problem.

For example, in a banking application, we can create a custom exception for insufficient balance.

```python
class InsufficientBalanceError(Exception):
    pass


balance = 1000
withdraw = 1500

try:
    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient balance")
    balance -= withdraw
    print("Withdrawal successful")
except InsufficientBalanceError as error:
    print(error)
```

## Implementation Explanation

I created `InsufficientBalanceError` as a custom exception for insufficient balance.

The `if` condition checks whether the withdrawal amount is greater than the available balance.

If the condition is true, I use `raise` to generate my custom exception.

The `except` block catches the custom exception and displays a meaningful message.

## Output

Insufficient balance

# ValueError

- `ValueError` occurs when a function receives a value of the correct type but an invalid value.
- It usually happens when Python cannot process the given value as expected.
- It is commonly seen when converting invalid strings into numbers.
- We can use `try` and `except` to handle this error safely.
  
For example, converting the string `"abc"` into an integer is not possible, so Python raises a `ValueError`.

```python
try:
    number = int("abc")
    print(number)
except ValueError:
    print("Invalid value. Please enter a valid number.")
```

## Implementation Explanation

I used `int()` to convert a string into an integer.

Since `"abc"` is not a valid number, Python raises a `ValueError`.

The `except` block catches the error and displays a meaningful message.

## Output

Invalid value. Please enter a valid number.

# TypeError

- `TypeError` occurs when an operation is performed on an inappropriate data type.
- It usually happens when two incompatible types are used together.
- For example, we cannot add a string and an integer directly.
- We can use `try` and `except` to handle this error.

For example, adding a string and an integer causes a `TypeError` because they are different and incompatible types.

```python
try:
    result = "10" + 5
    print(result)
except TypeError:
    print("Cannot add a string and an integer.")
```

## Implementation Explanation

I used a string and an integer in an addition operation.

Python cannot directly add these two different data types, so it raises a `TypeError`.

The `except` block catches the error and displays a meaningful message.

## Output

Cannot add a string and an integer.

# IndexError

- `IndexError` occurs when we try to access an index that does not exist.
- It commonly occurs when working with lists, tuples, or other indexed collections.
- The index must be within the valid range of the collection.
- We can use `try` and `except` to handle an invalid index safely.

For example, if a list has three items, its valid indexes are 0, 1, and 2. Trying to access index 3 causes an `IndexError`.

```python
numbers = [10, 20, 30]

try:
    print(numbers[3])
except IndexError:
    print("Index is outside the list range.")
```

## Implementation Explanation

I created a list containing three numbers.

The valid indexes are 0, 1, and 2, but I tried to access index 3.

This causes an `IndexError`, which is handled by the `except` block.

## Output

Index is outside the list range.

# KeyError

- `KeyError` occurs when we try to access a key that does not exist in a dictionary.
- A dictionary allows access to values using their keys.
- If the requested key is missing, Python raises a `KeyError`.
- We can use `try` and `except` to handle a missing key safely.

For example, if a dictionary contains a person's name and age, trying to access a key called `"city"` will cause a `KeyError` if that key does not exist.

```python
student = {
    "name": "Kesavarthini",
    "age": 22
}

try:
    print(student["city"])
except KeyError:
    print("The requested key does not exist.")
```

## Implementation Explanation

I created a dictionary containing the student's name and age.

I tried to access the `"city"` key, but that key is not present in the dictionary.

This causes a `KeyError`, which is handled by the `except` block.

## Output

The requested key does not exist.

# AttributeError

- `AttributeError` occurs when we try to access an attribute or method that an object does not have.
- It commonly happens when we use an incorrect method or attribute name.
- Python checks whether the object supports the requested attribute.
- We can use `try` and `except` to handle this error safely.
  
For example, a string has an `upper()` method, but it does not have a method called `uppercase()`.

```python
text = "hello"

try:
    print(text.uppercase())
except AttributeError:
    print("The object does not have this attribute or method.")
```

The string object does not have a method with that name, so Python raises an `AttributeError`.

The `except` block catches the error and displays a meaningful message.

## Output

The object does not have this attribute or method.

# Handling Multiple Exceptions

- A program can produce different types of exceptions.
- We can use multiple `except` blocks to handle different errors separately.
- Each `except` block can provide a specific response for a particular error.
- This makes the program easier to understand and handle different situations safely.

For example, a division program can produce a `ValueError` if the input is not a number or a `ZeroDivisionError` if the divisor is zero.

```python
try:
    number = int(input("Enter a number: "))
    divisor = int(input("Enter divisor: "))
    result = number / divisor
    print("Result:", result)

except ValueError:
    print("Please enter valid numbers.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## Implementation Explanation

I used two `except` blocks to handle two different errors.

`ValueError` handles invalid input, while `ZeroDivisionError` handles division by zero.

This makes the program respond differently depending on the type of error.

## Output

For valid input:

Result: 5.0

For invalid input:

Please enter valid numbers.

For zero as divisor:

Cannot divide by zero.
