# What is SQL?

- SQL stands for Structured Query Language.
- It is used to store, retrieve, update, and manage data in relational databases.
- SQL works with tables that contain rows and columns.
- SQL is commonly used in applications, data analysis, and AI/ML projects to work with structured data.

For example, a student database can contain a table with student names, ages, and marks.

We can use SQL to retrieve the students who have scored more than a particular mark.

```python
import sqlite3

connection = sqlite3.connect("students.db")

cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    marks INTEGER
)
""")

cursor.execute("""
INSERT INTO students (name, marks)
VALUES ('Anu', 85)
""")

connection.commit()

print("Student table created and data inserted.")
```

## Implementation Explanation

- `sqlite3` allows Python to work with an SQLite database.
- `connect()` creates or opens the database.
- `cursor()` is used to execute SQL commands.
- `CREATE TABLE` creates the students table.
- `INSERT INTO` adds student data to the table.
- `commit()` saves the changes to the database.

## Output

Student table created and data inserted.

## Implementation Explanation

- `sqlite3.connect()` connects Python to the database.

- `CREATE TABLE` creates a new table.
- `id`, `name`, and `marks` are the columns.

- `close()` closes the database connection after the operation.

## Output

Students table created successfully.

# INSERT Statement

# SELECT Statement

- The `SELECT` statement is used to retrieve data from a database table.
- We can select all columns or only specific columns.
- It is mainly used to view and analyze stored data.
- We can also use conditions with `SELECT` to retrieve specific records.

For example, we can retrieve all students from the `students` table.

We can also retrieve only the student names and marks.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM students")

students = cursor.fetchall()

for student in students:
    print(student)

connection.close()
```

## Implementation Explanation

- `SELECT *` retrieves all columns from the table.
- `fetchall()` retrieves all the records returned by the query.
- The `for` loop displays each student record.
- `SELECT` is useful when we need to read or analyze stored data.

## Output

The output displays the student records stored in the `students` table.

Example:

(1, 'Anu', 85)
(2, 'Priya', 90)

# WHERE Clause

- The `WHERE` clause is used to filter records based on a condition.
- It helps us retrieve only the records that match a specific requirement.
- We can use comparison operators such as `=`, `>`, `<`, `>=`, and `<=`.
- It is useful when we need specific data instead of all records.

## Demonstrate the Concept

For example, we can retrieve only the students who scored more than 85 marks.

Instead of displaying every student, the `WHERE` clause filters the required records.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT * FROM students
WHERE marks > 85
""")

students = cursor.fetchall()

for student in students:
    print(student)

connection.close()
```

## Implementation Explanation

- `WHERE` applies a condition to the query.
- `marks > 85` selects only students whose marks are greater than 85.
- `fetchall()` retrieves the matching records.
- The loop displays the filtered records.
- This is useful for finding specific data from a large table.

## Output

Example:

(2, 'Priya', 90)

Only students with marks greater than 85 are displayed.

# UPDATE Statement

- The `UPDATE` statement is used to modify existing records in a table.
- We can change one or more column values.
- The `WHERE` clause is usually used to identify which record should be updated.
- It is useful when existing information in a database needs to be corrected or changed.


For example, if a student's marks were entered incorrectly, we can update the marks using the student's ID.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
UPDATE students
SET marks = 95
WHERE name = 'Priya'
""")

connection.commit()

print("Student record updated successfully.")

connection.close()
```

## Implementation Explanation

- `UPDATE` is used to modify existing data.
- `SET` specifies the new value.
- `WHERE` identifies which record should be changed.
- `commit()` saves the updated data.
- Using `WHERE` is important because without it, all rows could be updated.

## Output

Student record updated successfully.

# DELETE Statement

- The `DELETE` statement is used to remove records from a table.
- The `WHERE` clause can be used to identify the record that should be deleted.
- It is useful when a record is no longer required or was entered incorrectly.
- We should use `WHERE` carefully because deleting without a condition can remove all records.


For example, we can delete a student record using the student's name.

Here, we will remove the record for `Priya`.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
DELETE FROM students
WHERE name = 'Priya'
""")

connection.commit()

print("Student record deleted successfully.")

connection.close()
```

## Implementation Explanation

- `DELETE FROM` is used to remove records from a table.
- `WHERE` identifies which record should be deleted.
- Here, the record where the name is `Priya` is removed.
- `commit()` saves the deletion.
- Using `WHERE` prevents accidentally deleting all records.

## Output

Student record deleted successfully.

# SQL Constraints

- Constraints are rules applied to table columns.
- They help maintain accurate and valid data in a database.
- Common constraints include `PRIMARY KEY`, `NOT NULL`, `UNIQUE`, `DEFAULT`, and `CHECK`.
- Constraints prevent invalid or duplicate data from being stored.

For example, a student ID can be a `PRIMARY KEY`, the student name can be `NOT NULL`, and marks can have a `CHECK` condition.

This ensures that the stored data follows the required rules.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS student_constraints (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT UNIQUE,
    marks INTEGER CHECK(marks >= 0 AND marks <= 100),
    status TEXT DEFAULT 'Active'
)
""")

print("Table with constraints created successfully.")

connection.close()
```

## Implementation Explanation

- `PRIMARY KEY` uniquely identifies each record.
- `NOT NULL` ensures that a value must be provided.
- `UNIQUE` prevents duplicate values.
- `CHECK` ensures that marks are between 0 and 100.
- `DEFAULT` automatically provides a value when no value is given.

## Output

Table with constraints created successfully.

# ORDER BY Clause

- The `ORDER BY` clause is used to sort records in a table.
- We can sort data in ascending (`ASC`) or descending (`DESC`) order.
- By default, `ORDER BY` uses ascending order.
- It is useful when we need to arrange data for easier analysis and viewing.


For example, we can sort students based on their marks.

Using `DESC` will display the student with the highest marks first.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT * FROM students
ORDER BY marks DESC
""")

students = cursor.fetchall()

for student in students:
    print(student)

connection.close()
```

## Implementation Explanation

- `ORDER BY marks` sorts the records based on the marks column.
- `DESC` sorts the records from highest to lowest.
- `fetchall()` retrieves the sorted records.
- The `for` loop displays each record in the sorted order.

## Output

Example:

(2, 'Priya', 95)
(1, 'Anu', 85)

The students are displayed from highest marks to lowest marks.

# Aggregate Functions

- Aggregate functions perform calculations on multiple rows.
- They return a single result from a group of records.
- Common aggregate functions are `COUNT()`, `SUM()`, `AVG()`, `MIN()`, and `MAX()`.
- They are useful for analyzing and summarizing data.

For example, we can calculate the number of students and find their average, highest, and lowest marks.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT
    COUNT(*) AS total_students,
    AVG(marks) AS average_marks,
    MIN(marks) AS lowest_marks,
    MAX(marks) AS highest_marks
FROM students
""")

result = cursor.fetchone()

print("Total students:", result[0])
print("Average marks:", result[1])
print("Lowest marks:", result[2])
print("Highest marks:", result[3])

connection.close()
```

## Implementation Explanation

- `COUNT()` counts the number of records.
- `AVG()` calculates the average value.
- `MIN()` finds the lowest value.
- `MAX()` finds the highest value.
- `fetchone()` retrieves the single result returned by the query.

## Output

Example:

Total students: 1
Average marks: 85.0
Lowest marks: 85
Highest marks: 85

# GROUP BY Clause

- The `GROUP BY` clause is used to group rows that have the same value.
- It is commonly used with aggregate functions like `COUNT()`, `SUM()`, and `AVG()`.
- It helps us summarize data based on a particular column.
- It is useful for analyzing data category-wise.

For example, students can be grouped based on their marks.

We can count how many students have each particular mark.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT marks, COUNT(*) AS student_count
FROM students
GROUP BY marks
""")

results = cursor.fetchall()

for result in results:
    print("Marks:", result[0], "Students:", result[1])

connection.close()
```

## Implementation Explanation

- `GROUP BY marks` groups students having the same marks.
- `COUNT(*)` counts the number of students in each group.
- `fetchall()` retrieves all grouped results.
- The `for` loop displays each group.
- `GROUP BY` is useful for summarizing data based on categories.

## Output

Example:

Marks: 85 Students: 1

The output shows the number of students for each marks group.

# HAVING Clause

- The `HAVING` clause is used to filter grouped results.
- It is commonly used with `GROUP BY` and aggregate functions.
- `WHERE` filters individual rows, while `HAVING` filters groups.
- It is useful when we want to apply conditions to calculated or grouped results.

For example, we can group students based on their marks and display only the marks that have at least one student.

The `HAVING` clause filters the grouped result.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT marks, COUNT(*) AS student_count
FROM students
GROUP BY marks
HAVING COUNT(*) >= 1
""")

results = cursor.fetchall()

for result in results:
    print("Marks:", result[0], "Students:", result[1])

connection.close()
```

## Implementation Explanation

- `GROUP BY marks` groups students based on their marks.
- `COUNT(*)` counts students in each group.
- `HAVING COUNT(*) >= 1` filters the grouped results.
- Unlike `WHERE`, `HAVING` is used after grouping.
- It is useful when filtering results based on aggregate calculations.

## Output

Example:

Marks: 85 Students: 1

The output contains only groups that satisfy the HAVING condition.

# DISTINCT

- `DISTINCT` is used to remove duplicate values from query results.
- It returns only unique values.
- It is useful when we want to find different or unique values in a column.
- It does not delete data from the table; it only removes duplicates from the displayed result.

For example, if multiple students have the same marks, we can use `DISTINCT` to display each marks value only once.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT DISTINCT marks
FROM students
""")

results = cursor.fetchall()

for result in results:
    print("Unique marks:", result[0])

connection.close()
```

## Implementation Explanation

- `DISTINCT` returns only unique values.
- `SELECT DISTINCT marks` retrieves each different marks value only once.
- `fetchall()` retrieves all unique results.
- The `for` loop displays the unique values.
- `DISTINCT` does not change or delete the original data.

## Output

Example:

Unique marks: 85

Only unique marks values are displayed.

- It is commonly used when retrieving the top few records from a table.

For example, we can retrieve only the first two students from the `students` table.

# SQL JOINs

- A JOIN is used to combine data from two or more related tables.
- JOINs are useful when required information is stored in different tables.
- Common JOIN types include INNER JOIN, LEFT JOIN, RIGHT JOIN, and FULL JOIN.
- JOINs are commonly used in real-world applications and data analysis.

For example, one table can store student details and another table can store their course details.

We can use a JOIN to combine the student name with their course.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS courses (
    student_id INTEGER,
    course TEXT
)
""")

cursor.execute("""
INSERT INTO courses (student_id, course)
VALUES (1, 'Python')
""")

connection.commit()

cursor.execute("""
SELECT students.name, courses.course
FROM students
INNER JOIN courses
ON students.id = courses.student_id
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `INNER JOIN` combines matching records from both tables.
- `students.id` and `courses.student_id` are used to connect the tables.
- The `ON` condition defines how the tables are related.
- `SELECT` retrieves the student's name and course.
- JOINs are useful when related information is stored in separate tables.

## Output

Example:

('Anu', 'Python')

The output combines data from the `students` and `courses` tables.

# LEFT JOIN

- A `LEFT JOIN` combines data from two tables.
- It returns all records from the left table.
- If there is no matching record in the right table, the result contains `NULL` for the right table's columns.
- It is useful when we want to keep all records from the main table even if related data is missing.


For example, we can display all students along with their courses.

Even if a student does not have a course, the student will still appear in the result.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students.name, courses.course
FROM students
LEFT JOIN courses
ON students.id = courses.student_id
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `students` is the left table.
- `courses` is the right table.
- `LEFT JOIN` keeps all records from the `students` table.
- The `ON` condition matches the student ID with the course student ID.
- If a student has no matching course, the course value will be `NULL`.

# RIGHT JOIN

- A `RIGHT JOIN` returns all records from the right table.
- It also returns matching records from the left table.
- If there is no matching record in the left table, the result contains `NULL` for the left table's columns.
- It is useful when the right table contains the records we want to keep completely.


For example, we can display all courses along with student names.

Even if a course does not have a matching student, the course can still appear in the result.

```python
cursor.execute("""
SELECT students.name, courses.course
FROM students
RIGHT JOIN courses
ON students.id = courses.student_id
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `courses` is the right table in this query.
- `RIGHT JOIN` keeps all records from the `courses` table.
- The `ON` condition matches `students.id` with `courses.student_id`.
- If a course has no matching student, the student name will be `NULL`.
- RIGHT JOIN is useful when all records from the right table must be retained.

# FULL JOIN

- A `FULL JOIN` returns all records from both tables.
- It includes matching records from both tables.
- If there is no matching record, `NULL` is returned for the missing side.
- It is useful when we want to see all records from two related tables.

For example, we can combine the `students` and `courses` tables.

The result can contain students without courses and courses without matching students.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students.name, courses.course
FROM students
FULL OUTER JOIN courses
ON students.id = courses.student_id
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `FULL OUTER JOIN` combines records from both tables.
- Matching student and course records are combined.
- If a student has no matching course, the course value is `NULL`.
- If a course has no matching student, the student name is `NULL`.
- This allows us to see all records from both tables.

## Output

Example:

('Anu', 'Python')
('Ravi', None)
(None, 'Java')

The exact output depends on the data stored in the tables.

# CROSS JOIN

- A `CROSS JOIN` combines every row from the first table with every row from the second table.
- It produces all possible combinations between the two tables.
- It does not require an `ON` condition.
- It is useful when we need to generate combinations of values from two tables.


For example, if there are 2 students and 2 courses, a CROSS JOIN produces 4 combinations.

Each student is combined with every course.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students.name, courses.course
FROM students
CROSS JOIN courses
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `CROSS JOIN` combines every student with every course.
- No `ON` condition is required.
- If there are 2 students and 2 courses, the result contains 4 combinations.
- It is useful when we need all possible combinations between two sets of data.

# UNION

- `UNION` is used to combine the results of two or more `SELECT` queries.
- The queries should return the same number of columns with compatible data types.
- `UNION` removes duplicate rows from the final result.
- It is useful when we want to combine similar data from different queries or tables.

For example, we can combine student names and course names into one result.

Both queries return one column, so their results can be combined using `UNION`.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name AS value
FROM students

UNION

SELECT course AS value
FROM courses
""")

results = cursor.fetchall()

for result in results:
    print(result[0])

connection.close()
```

## Implementation Explanation

- The first `SELECT` retrieves student names.
- The second `SELECT` retrieves course names.
- `UNION` combines the results of both queries.
- Both queries return one column, so they can be combined.
- Duplicate values are removed automatically by `UNION`.

# CASE Statement

- The `CASE` statement is used to apply conditional logic in SQL.
- It works similar to `if-elif-else` in Python.
- It checks conditions and returns a value based on the first matching condition.
- It is useful for categorizing or creating new values from existing data.

For example, we can categorize students based on their marks.

- Marks 90 or above → Excellent
- Marks 75 or above → Good
- Otherwise → Needs Improvement

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, marks,
    CASE
        WHEN marks >= 90 THEN 'Excellent'
        WHEN marks >= 75 THEN 'Good'
        ELSE 'Needs Improvement'
    END AS performance
FROM students
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `CASE` checks conditions one by one.
- `WHEN` defines a condition.
- `THEN` specifies the value returned when the condition is true.
- `ELSE` provides a value when none of the conditions are true.
- `END` marks the end of the CASE expression.

# Subqueries

- A subquery is a SQL query written inside another SQL query.
- The inner query is executed first and its result is used by the outer query.
- Subqueries are useful when one query depends on the result of another query.
- They are commonly used with `SELECT`, `WHERE`, and aggregate functions.


For example, we can first calculate the average marks of all students.

Then, using the outer query, we can find students who scored higher than that average.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, marks
FROM students
WHERE marks > (
    SELECT AVG(marks)
    FROM students
)
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- The inner query calculates the average marks using `AVG()`.
- The outer query retrieves the student name and marks.
- The `WHERE` condition compares each student's marks with the average.
- Only students who scored higher than the average are returned.
- The inner query is called the subquery.

# Views

- A view is a virtual table created from the result of a SQL query.
- It does not normally store the actual data separately.
- Views can simplify complex queries and make data easier to access.
- They are useful when the same query needs to be used repeatedly.

For example, we can create a view that contains only students who scored more than 80 marks.

We can then query the view like a normal table.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
CREATE VIEW IF NOT EXISTS high_scorers AS
SELECT name, marks
FROM students
WHERE marks > 80
""")

cursor.execute("SELECT * FROM high_scorers")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `CREATE VIEW` creates a virtual table based on a SQL query.
- `high_scorers` is the name of the view.
- The view contains students whose marks are greater than 80.
- `SELECT * FROM high_scorers` retrieves data from the view.
- Views are useful for simplifying frequently used queries.

# Indexes

- An index is used to make searching and retrieving data faster.
- It creates a separate structure that helps SQL find rows efficiently.
- Indexes are useful when a table contains a large amount of data.
- They are commonly created on columns that are frequently used in `WHERE`, `JOIN`, or `ORDER BY` conditions.

For example, if we frequently search students using their names, we can create an index on the `name` column.

This helps the database find matching records more efficiently.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
CREATE INDEX IF NOT EXISTS idx_student_name
ON students(name)
""")

cursor.execute("""
SELECT name, marks
FROM students
WHERE name = 'Anu'
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `CREATE INDEX` creates an index on the `name` column.
- `idx_student_name` is the name given to the index.
- The index helps improve searching based on the student name.
- `SELECT` retrieves the student record using the indexed column.
- Indexes can improve query performance, especially for large tables.

# ROW_NUMBER()

- `ROW_NUMBER()` is a SQL window function that assigns a unique number to each row.
- The numbering starts from 1.
- We can use `ORDER BY` to decide the order of numbering.
- It is useful for ranking rows, pagination, and identifying individual records.

For example, we can number students based on their marks from highest to lowest.

The student with the highest mark will get row number 1.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT
    name,
    marks,
    ROW_NUMBER() OVER (ORDER BY marks DESC) AS row_number
FROM students
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `ROW_NUMBER()` assigns a unique number to each row.
- `OVER()` defines how the window function should work.
- `ORDER BY marks DESC` sorts students from highest marks to lowest.
- The result gives each student a different row number.
- This is useful when we need a unique sequence for records.

# RANK()

- `RANK()` is a SQL window function used to assign a rank to each row.
- Rows with the same value get the same rank.
- When two or more rows have the same rank, the next rank is skipped.
- It is useful for ranking students, employees, products, or other records based on a value.


For example, we can rank students based on their marks from highest to lowest.

If two students have the same marks, they will receive the same rank.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT
    name,
    marks,
    RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM students
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `RANK()` assigns a rank based on the marks.

# DENSE_RANK()

- `DENSE_RANK()` is a SQL window function used to rank rows based on a value.
- Rows with the same value receive the same rank.
- Unlike `RANK()`, it does not skip the next rank when there is a tie.
- It is useful when we want continuous rankings.

For example, we can rank students based on their marks from highest to lowest.

If two students have the same marks, they get the same rank, and the next student gets the next continuous rank.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT
    name,
    marks,
    DENSE_RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM students
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `DENSE_RANK()` assigns a rank based on the marks.
- `OVER()` defines the ranking window.
- `ORDER BY marks DESC` ranks students from highest to lowest marks.
- Students with the same marks receive the same rank.
- Unlike `RANK()`, the next rank is not skipped.

# LAG()

- `LAG()` is a SQL window function used to access a value from a previous row.
- It allows us to compare the current row with an earlier row.
- We can use `ORDER BY` to decide the order of the rows.
- It is useful for comparing marks, sales, prices, or other values between rows.


For example, we can display each student's marks along with the marks of the previous student.

This helps us compare the current value with the previous value.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT
    name,
    marks,
    LAG(marks) OVER (ORDER BY id) AS previous_marks
FROM students
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `LAG(marks)` gets the marks from the previous row.
- `OVER()` defines the window for the function.
- `ORDER BY id` decides the order in which rows are compared.
- The first row has no previous row, so its previous value will be `None`.
- This is useful for comparing current and previous values.

# LEAD()

- `LEAD()` is a SQL window function used to access a value from the next row.
- It allows us to compare the current row with a later row.
- We can use `ORDER BY` to decide the order of the rows.
- It is useful for comparing current and future values in data analysis.

For example, we can display each student's marks along with the marks of the next student.

This helps us compare the current value with the next value.

```python
import sqlite3

connection = sqlite3.connect("students.db")
cursor = connection.cursor()

cursor.execute("""
SELECT
    name,
    marks,
    LEAD(marks) OVER (ORDER BY id) AS next_marks
FROM students
""")

results = cursor.fetchall()

for result in results:
    print(result)

connection.close()
```

## Implementation Explanation

- `LEAD(marks)` gets the marks from the next row.
- `OVER()` defines the window for the function.
- `ORDER BY id` decides the order in which rows are processed.
- The last row has no next row, so its next value will be `None`.
- This is useful when comparing the current value with a future value.
